## Setting the path to input files

```python
import os
from singscore.singscore import *

BASE_DIR = os.getcwd()   # current working directory
print(BASE_DIR)

entrez_sample_path = os.path.normpath('{}/{}'.format(BASE_DIR,
                                'singscore/test/test_data/entrez_sample.txt'))

sig_path = os.path.normpath('{}/{}'.format(BASE_DIR,
                                'singscore/test/test_sigs/tgfb_upDown.txt'))
```

## Reading the data

```python
data = pandas.read_csv(open(entrez_sample_path, 'r'), header=
'infer', sep='\t')
data = data.set_index(keys='EntrezID')
data
```

# prepare signatures

### seperating the aanotated up and downregulated genes

```python
sigs = pandas.read_csv(open(sig_path, 'r'), header=
'infer', sep='\t')
sigs
```

```python
# subset the data for up and down
up = sigs[sigs['upDown'] == 'up']
down = sigs[sigs['upDown'] == 'down']
up
```

```python
# get a list of ids
up = list(up['EntrezID'])
down = list(down['EntrezID'])
up
```

```python
print(f'{len(up), len(down)}')
```

```python
up_present = [gene for gene in up if gene in data.index]
down_present = [gene for gene in down if gene in data.index]

# Print counts and summary
print(f"Total up genes in signature: {len(up)}")
print(f"Up genes found in expression data: {len(up_present)}")

print(f"Total down genes in signature: {len(down)}")
print(f"Down genes found in expression data: {len(down_present)}")

# Optionally, print the missing ones
missing_up = set(up) - set(data.index)
missing_down = set(down) - set(data.index)

print(f"Missing up genes: {len(missing_up)}")
print(f"Missing down genes: {len(missing_down)}")
```

# Scoring the genes

```python
scored_data = score(up_gene=up, down_gene=down, sample=data,
                    norm_method='theoretical', full_data=True)
print(scored_data)
```

```python
# plot dispersion of scores
output_path = os.path.normpath('{}/{}'.format(BASE_DIR,
                                    'singscore/test/output_JN/dispersion.png'))

plotdispersion(scored_data, ctrlstring='Ctrl', teststring='TGFb',
               testlabel='TGFb', colour_1='g', colour_2='b',
               outpath=output_path, show=False)
```

```python
output_path = os.path.normpath('{}/{}'.format(BASE_DIR,
                                        'singscore/test/output_JN/nulldist.png'))
permd = permutate(data, n_up=193, n_down=108)
pvals = empiricalpval(permutations=permd, score=scored_data)
print(pvals)
```

```python
nulldistribution(permutations=permd, score=scored_data, nrows=2, ncols=5,
                 threshold=0.05, outpath=output_path, show=False)
```
